```python

file_name = "__2403 01_240710.xlsx"
full_path = f"{data_base_dir}\\{file_name}"
df = pd.read_excel(full_path, sheet_name=0, engine="openpyxl")  
for i, col in enumerate(df.columns):
    unique_values = df[col].unique()
    values_str = ", ".join(map(str, unique_values[:4]))
    print(f"{i+1}\t{col}\t{df[col].dtype}\t{len(unique_values)}\t{values_str}")
```
